```python
import matplotlib.pyplot as plt

sms = [1, 2, 4, 8, 16, 32, 64]
file_2_1_misses = [406791, 348340, 350914, 355851, 361646, 371062, 388049]
file_2_2_misses = [522516, 409340, 399863, 408931, 416731, 427684, 447234]
file_2_3_misses = [962271, 858005, 879372, 814100, 719262, 770463, 837214]

plt.figure(figsize=(10, 6))

plt.plot(sms, file_2_1_misses, marker='o', label='file2.1', linewidth=2)
plt.plot(sms, file_2_2_misses, marker='s', label='file2.2', linewidth=2)
plt.plot(sms, file_2_3_misses, marker='^', label='file2.3', linewidth=2)

plt.xscale('log', base=2)
plt.xticks(sms, sms)

plt.xlabel('Number of SMs', fontsize=12)
plt.ylabel('Number of L1 Cache Misses', fontsize=12)
plt.title('L1 Cache Misses vs. Number of SMs', fontsize=14, pad=15)
plt.legend(title='App/File', fontsize=10)
plt.grid(True, which='both', alpha=0.5)

plt.show()
```
